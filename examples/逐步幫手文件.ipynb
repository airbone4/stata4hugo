````
%%stata
use "https://rmilab.nkust.edu.tw/files/bookdata/lecture/chen_ex_10_2.dta", clear
sw reg y x1-x4 , pe(0.99) pr(0.999) forward
reg y x1
reg y x2
reg y x3
reg y x4
//x2 被選,因為F最大(紅字)
reg y x2 x1
reg y x2 x3
reg y x2 x4
// adding x4 (F最大，見紅圈）
// 然後對目前的兩個變數(x2,x4)塞選,兩個變數的p值,不用被移出因為沒有大於0.999
reg y x2 x4 x1
 

````
